# Veyraxis Sentinel: Exploratory Data Analysis & Model Prototyping

This notebook demonstrates:
1. Dataset integrity inspection and class distribution analysis.
2. Bounding box visualizations and augmentations.
3. Real-time inference using the `VeyraxisDetector` and `InferenceEngine`.

```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt

from veyraxis.data.validator import DatasetValidator
from veyraxis.inference.engine import InferenceEngine
from veyraxis.inference.visualizer import Visualizer
```

## 1. Run Dataset Validation

```python
classes = ['person', 'vehicle', 'helmet', 'fire', 'damaged_component']
validator = DatasetValidator(dataset_root=Path('../dataset'), classes=classes)
report = validator.validate(output_dir=Path('../outputs/notebook_reports'))
print(f"Total Images: {report.total_images}, Total Annotations: {report.total_annotations}")
```

## 2. Load Model & Run Live Inference

```python
engine = InferenceEngine(model_path='../runs/checkpoints/best.pt', device='auto', conf_threshold=0.25)
sample_path = Path('../dataset/images/test').glob('*.jpg')
first_image = next(sample_path, None)

if first_image:
    img = cv2.imread(str(first_image))
    result = engine.predict_image(img, image_id=first_image.name)
    print(f"Latency: {result.latency_ms} ms | Detections: {len(result.detections)}")
    vis = Visualizer().draw(img, result)
    plt.figure(figsize=(8, 8))
    plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()
```
